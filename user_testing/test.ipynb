```python
import json
import subprocess

import requests

# Define the request payload
payload = {
    "contents": [
        {
            "role": "user",
            "parts": [
                {"text": "What is the meaning of life?"},
            ],
        }
    ],
    "generationConfig": {
        "maxOutputTokens": 8192,
        "temperature": 0,
        "topP": 0.95,
    },
}

# Project and API details
PROJECT_ID = "autogenai-8ed11"
LOCATION_ID = "us-central1"
API_ENDPOINT = "us-central1-aiplatform.googleapis.com"
MODEL_ID = "gemini-experimental"

# Construct the URL
url = f"https://{API_ENDPOINT}/v1/projects/{PROJECT_ID}/locations/{LOCATION_ID}/publishers/google/models/{MODEL_ID}:streamGenerateContent"


# Get the access token
def get_access_token():
    token = subprocess.check_output("gcloud auth print-access-token", shell=True).decode().strip()
    return token


# Set up headers
headers = {"Content-Type": "application/json", "Authorization": f"Bearer {get_access_token()}"}

# Make the POST request
response = requests.post(url, headers=headers, data=json.dumps(payload))

# Print the response
text_response = "".join(
    [line["candidates"][0]["content"]["parts"][0]["text"] for line in response.json()]
)
print(text_response)
```

```python
def google_message_parse(response: dict):
    try:
        if isinstance(response, dict):
            text_response = response["candidates"][0]["content"]["parts"][0]["text"]
        else:
            text_response = "".join(
                [
                    line["candidates"][0]["content"]["parts"][0]["text"]
                    for line in response
                    if "candidates" in line and "content" in line["candidates"][0]
                ]
            )
        return text_response
    except Exception as e:
        if response in [{}, []]:
            return "None"
        else:
            raise ValueError(
                f"Response does not contain the expected key 'candidates'. Error: {e}. Response: {response}"
            )
```
